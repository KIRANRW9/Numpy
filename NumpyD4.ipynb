#Arithmatic operators

```python
import numpy as np
g=np.array([[2,3,4],[5,6,7]])
h=np.array([[2.1,3.5,2.2],[1.2,5.3,2]])

#Arithmatic opeerations
print("g+h\n",g+h)
print()
print(np.add(g,h))

print("h-g:\n",h-g)
print()
print(np.subtract(h,g))

print("g*h:\n",g*h)
print()
print(np.multiply(g,h))

print("g/h:\n",g/h)
print()
print(np.divide(g,h))

print("g//h:\n",g//h)
print()
print(np.floor_divide(g,h))
```

#a.view() → Creates a view (linked copy)
1.Copying Arrays in NumPy
When you work with arrays in NumPy, you sometimes want to create a new array that is either:

Connected to the original (called a "view")

Separate and independent (called a "copy")

```python
a=np.arange(1,9)
h=a.view()
h[[0,1]]=12,13
print(a)
print(h)
```

#🔹 a.copy() or np.copy(a) → Creates a deep copy
python
Copy
Edit


```python
x=a.copy()
y=np.copy(a)
y[[0,1]]=9,4
print(x)
print(y)
```

 2.Saving & Loading Arrays
 np.save('file', array) → saves to .npy file

```python
np.save("myar_a",x)

loaded=np.load("myar_a.npy")
print(loaded)
```

```python
#To save multiple arrays
np.savez("arrays.npz",x,y)
l1=np.load("arrays.npz")
print(l1['arr_0'])
print("______________")
print(l1['arr_1'])
```
