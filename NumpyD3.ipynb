#Inspecting Your Array


1.a.shape — Array Dimensions

Tells you the size of the array in each dimension (rows, columns, etc.).

Returns a tuple.

```python
import numpy as np
a=np.array([[1,2,3],[4,5,6]])
print(a.shape)
```

2. len(a) — Length of the Array
Returns the number of rows (or the size of the first dimension).

```python
print(len(a))
```

3.b.ndim — Number of Dimensions
Returns the number of axes or dimensions.

1D → vector, 2D → matrix, 3D → cube, etc.



```python
a.ndim
```

4.e.size — Total Number of Elements
Returns how many elements are in the array.

```python
print(a.size)
```

5.b.dtype — Data Type of Elements
Shows the actual NumPy data type of the array elements.

```python
a.dtype
```

b.dtype.name — Name of Data Type
Returns the name of the data type in string format.

```python
a.dtype.name
```

b.astype(int) — Type Conversion
Converts the array into another data type.

Returns a new array with the changed type

```python
floattype_array=np.array([1.2,3.4,5.3])
print(floattype_array)
inttype_array=floattype_array.astype(int)
print(inttype_array)
```
