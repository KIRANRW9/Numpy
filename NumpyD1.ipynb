***NumPy***

The NumPy library is the core library for scientific computing in
Python. It provides a high-performance multidimensional array
object, and tools for working with these arrays.

1.**Creating Arrays** of 1D,2D and 3D.






```python
import numpy as np
#1D Array
a=np.array([1,2,3])
print(a)

print("----------------------------------------------------")
#2D Array
b=np.array([[1.9,2,3],[4,5,6]],dtype=float)
print(b)

print("----------------------------------------------------")
#3D Arrray
c=np.array([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]],dtype=float)
print(c)
```

2.This creates 3X4 Array filled with 0.

```python
a=np.zeros((3,4))
print(a)
```

3.This creates 3D array (5 Blocks of 3X4) filled with ones using integer type.

```python
b=np.ones((5,3,4),dtype=np.int16)
print(b)
```

3.Array with step value
# Starts at 10 and goes upto 25 but 25 in not include.

```python
c=np.arange(10,25,2)
print(c)
```

4.Array with equal intervals
It divides the 1-50 into 5 equal parts

```python
d=np.linspace(1,50,5)
print(d)
```

5.Constant Matrix it creates a 3X5 matrix with full of 7.

```python
e=np.full((3,5),7)
print(e)
```

6.It creates a 5X5 Identity Matrix (1's on Diagonl 0's else where

```python
f=np.eye(5)
print(f)
```

7.Random values from 0 to 1 float values of 4X5 matrix

```python
g=np.random.random((4,5))
print(g)
```

8.Creates 3X2 Array without Initializing the values
